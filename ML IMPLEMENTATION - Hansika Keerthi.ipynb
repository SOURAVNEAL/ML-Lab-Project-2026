```python

# ============================================================
# SECTION 09: VALIDATE IMAGES WITH RESUMABLE PROGRESS
# Purpose: Identify unreadable images before feature extraction.
# All scan records are stored in Google Drive.
# ============================================================

import os
from PIL import Image

DATASET_DIR = "/content/drive/MyDrive/crop_disease_dataset"
RESULTS_DIR = "/content/drive/MyDrive/CropDiseaseProject/results"
os.makedirs(RESULTS_DIR, exist_ok=True)

VALID_LOG = os.path.join(RESULTS_DIR, "valid_images.txt")
BAD_LOG = os.path.join(RESULTS_DIR, "corrupted_images.txt")

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

# Collect image paths in a consistent order.
image_paths = []
for root, _, filenames in os.walk(DATASET_DIR):
    for filename in filenames:
        if os.path.splitext(filename)[1].lower() in IMAGE_EXTENSIONS:
            image_paths.append(os.path.join(root, filename))

image_paths.sort()
print("Total image files found:", len(image_paths))

# Load prior scan results to support resuming.
valid_paths = set()
bad_paths = set()

for log_path, destination in [
    (VALID_LOG, valid_paths),
    (BAD_LOG, bad_paths),
]:
    if os.path.exists(log_path):
        with open(log_path, "r", encoding="utf-8") as f:
            destination.update(line.strip() for line in f if line.strip())

checked_paths = valid_paths | bad_paths
print("Previously checked:", len(checked_paths))

# Append results in batches to preserve progress.
with open(VALID_LOG, "a", encoding="utf-8") as valid_file, \
     open(BAD_LOG, "a", encoding="utf-8") as bad_file:

    pending = 0

    for index, path in enumerate(image_paths, start=1):
        if path in checked_paths:
            continue

        try:
            with Image.open(path) as img:
                img.verify()
            valid_file.write(path + "\n")
            valid_paths.add(path)
        except Exception:
            bad_file.write(path + "\n")
            bad_paths.add(path)

        pending += 1

        if pending >= 500:
            valid_file.flush()
            bad_file.flush()
            pending = 0
            print(
                f"Progress: {len(valid_paths) + len(bad_paths)}"
                f"/{len(image_paths)} checked"
            )

    valid_file.flush()
    bad_file.flush()

print("\nImage validation completed.")
print("Readable images:", len(valid_paths))
print("Unreadable images:", len(bad_paths))
print("Validation records saved in:", RESULTS_DIR)
```

```python

# ============================================================
# SECTION 11A: INITIALIZE RESNET50 FEATURE EXTRACTION
# Purpose: Load pretrained ResNet50 and prepare checkpoint paths.
# Output: A 2048-dimensional feature vector per image.
# ============================================================

import os
import json
import numpy as np
import tensorflow as tf
from tensorflow.keras.applications.resnet50 import ResNet50

# Dataset and persistent storage paths
DATASET_DIR = "/content/drive/MyDrive/crop_disease_dataset"
PROJECT_DIR = "/content/drive/MyDrive/CropDiseaseProject"
FEATURE_DIR = os.path.join(PROJECT_DIR, "features")
CHECKPOINT_DIR = os.path.join(PROJECT_DIR, "checkpoints")
RESULTS_DIR = os.path.join(PROJECT_DIR, "results")

for folder in [FEATURE_DIR, CHECKPOINT_DIR, RESULTS_DIR]:
    os.makedirs(folder, exist_ok=True)

VALID_LOG = os.path.join(RESULTS_DIR, "valid_images.txt")
BAD_LOG = os.path.join(RESULTS_DIR, "corrupted_images.txt")

# Read only the images already marked readable during validation.
with open(VALID_LOG, "r", encoding="utf-8") as f:
    valid_image_paths = [line.strip() for line in f if line.strip()]

print("Readable image paths loaded:", len(valid_image_paths))

# Extract class labels from the top-level folder names.
class_names = sorted(
    name for name in os.listdir(DATASET_DIR)
    if os.path.isdir(os.path.join(DATASET_DIR, name))
)

class_to_index = {name: idx for idx, name in enumerate(class_names)}

# Confirm every image belongs to a known class folder.
for path in valid_image_paths:
    relative_path = os.path.relpath(path, DATASET_DIR)
    class_name = relative_path.split(os.sep)[0]

    if class_name not in class_to_index:
        raise ValueError(f"Unexpected class folder in path: {path}")

# Load pretrained ResNet50 without its classification head.
# pooling="avg" produces a 2048-dimensional feature vector.
base_model = ResNet50(
    weights="imagenet",
    include_top=False,
    pooling="avg",
    input_shape=(224, 224, 3)
)

base_model.trainable = False

print("ResNet50 loaded successfully.")
print("Number of classes:", len(class_names))
print("Feature dimension:", base_model.output_shape[-1])
```

```python

# ============================================================
# SECTION 11B: RESUME FEATURE EXTRACTION AFTER IMAGE ERROR
# Purpose:
# 1. Preserve previously completed batches.
# 2. Skip batches that already have completion markers.
# 3. Skip images that cannot be decoded.
# 4. Save each new batch and its labels to Google Drive.
# ============================================================

import os
import json
import numpy as np
import tensorflow as tf
from PIL import Image
from tensorflow.keras.applications.resnet50 import preprocess_input

BATCH_SIZE = 32
IMAGE_SIZE = (224, 224)

FEATURE_DIR = "/content/drive/MyDrive/CropDiseaseProject/features"
CHECKPOINT_DIR = "/content/drive/MyDrive/CropDiseaseProject/checkpoints"
RESULTS_DIR = "/content/drive/MyDrive/CropDiseaseProject/results"

os.makedirs(FEATURE_DIR, exist_ok=True)
os.makedirs(CHECKPOINT_DIR, exist_ok=True)
os.makedirs(RESULTS_DIR, exist_ok=True)

# Load the exact image list used by the original extraction run.
manifest_path = os.path.join(CHECKPOINT_DIR, "feature_manifest.json")

with open(manifest_path, "r", encoding="utf-8") as f:
    manifest = json.load(f)

paths = manifest["image_paths"]
labels_all = manifest["labels"]

assert len(paths) == len(labels_all), "Image and label counts differ!"

total_images = len(paths)
total_batches = (total_images + BATCH_SIZE - 1) // BATCH_SIZE

# Record newly discovered unreadable images separately.
failed_log = os.path.join(RESULTS_DIR, "extraction_failed_images.txt")

print("Total images in original manifest:", total_images)
print("Total batches:", total_batches)
print("Previously completed batches will be skipped.")

for batch_index, start in enumerate(range(0, total_images, BATCH_SIZE)):
    end = min(start + BATCH_SIZE, total_images)

    feature_path = os.path.join(
        FEATURE_DIR, f"features_{batch_index:05d}.npy"
    )
    label_path = os.path.join(
        FEATURE_DIR, f"labels_{batch_index:05d}.npy"
    )
    marker_path = os.path.join(
        CHECKPOINT_DIR, f"batch_{batch_index:05d}.json"
    )
    paths_path = os.path.join(
        FEATURE_DIR, f"paths_{batch_index:05d}.json"
    )

    # Skip a batch only when its files and marker all exist.
    if all(os.path.isfile(p) for p in [
        feature_path, label_path, marker_path, paths_path
    ]):
        continue

    batch_features_input = []
    batch_labels = []
    batch_paths = []
    failed_paths = []

    # Decode images individually so one damaged image cannot
    # interrupt the entire batch.
    for path, label in zip(paths[start:end], labels_all[start:end]):
        try:
            with Image.open(path) as image:
                image = image.convert("RGB")
                image.load()
                image = image.resize(IMAGE_SIZE)
                array = np.asarray(image, dtype=np.float32)

            batch_features_input.append(array)
            batch_labels.append(label)
            batch_paths.append(path)

        except Exception as exc:
            failed_paths.append(path)
            print(f"Skipping unreadable image: {path}")
            print(f"Reason: {type(exc).__name__}: {exc}")

    # Run inference only if this batch contains readable images.
    if batch_features_input:
        images = preprocess_input(
            np.stack(batch_features_input, axis=0)
        )

        features = base_model.predict(
            images, batch_size=BATCH_SIZE, verbose=0
        ).astype(np.float32)
    else:
        features = np.empty((0, 2048), dtype=np.float32)

    labels_array = np.asarray(batch_labels, dtype=np.int32)

    # Save batch data to temporary files before finalizing.
    temp_feature = feature_path + ".tmp.npy"
    temp_label = label_path + ".tmp.npy"
    temp_paths = paths_path + ".tmp"

    np.save(temp_feature, features)
    np.save(temp_label, labels_array)

    with open(temp_paths, "w", encoding="utf-8") as f:
        json.dump(batch_paths, f)

    os.replace(temp_feature, feature_path)
    os.replace(temp_label, label_path)
    os.replace(temp_paths, paths_path)

    # Save the marker last, after the batch files are in place.
    with open(marker_path, "w", encoding="utf-8") as f:
        json.dump({
            "batch_index": batch_index,
            "start": start,
            "end": end,
            "readable_count": len(batch_paths),
            "failed_count": len(failed_paths),
            "feature_shape": list(features.shape)
        }, f)

    # Record failures for later review.
    if failed_paths:
        with open(failed_log, "a", encoding="utf-8") as f:
            for path in failed_paths:
                f.write(path + "\n")

    print(
        f"Batch {batch_index + 1}/{total_batches} complete | "
        f"usable: {len(batch_paths)} | "
        f"skipped: {len(failed_paths)}"
    )

print("\nResume pass finished.")
print("Feature files:", FEATURE_DIR)
print("Checkpoint files:", CHECKPOINT_DIR)
print("New failure log:", failed_log)
```

```python

# ============================================================
# SECTION 12: VERIFY FEATURE EXTRACTION CHECKPOINTS
# Purpose: Check all saved batches before combining the data.
# ============================================================

import os
import json
import numpy as np

FEATURE_DIR = "/content/drive/MyDrive/CropDiseaseProject/features"
CHECKPOINT_DIR = "/content/drive/MyDrive/CropDiseaseProject/checkpoints"

with open(
    os.path.join(CHECKPOINT_DIR, "feature_manifest.json"),
    "r",
    encoding="utf-8"
) as f:
    manifest = json.load(f)

total_images = len(manifest["image_paths"])
batch_size = manifest["batch_size"]
total_batches = (total_images + batch_size - 1) // batch_size

missing_batches = []
invalid_batches = []
total_saved_images = 0

for i in range(total_batches):
    feature_path = os.path.join(FEATURE_DIR, f"features_{i:05d}.npy")
    label_path = os.path.join(FEATURE_DIR, f"labels_{i:05d}.npy")
    paths_path = os.path.join(FEATURE_DIR, f"paths_{i:05d}.json")
    marker_path = os.path.join(CHECKPOINT_DIR, f"batch_{i:05d}.json")

    if not all(os.path.isfile(p) for p in [
        feature_path, label_path, paths_path, marker_path
    ]):
        missing_batches.append(i)
        continue

    try:
        features = np.load(feature_path, mmap_mode="r")
        labels = np.load(label_path, mmap_mode="r")

        with open(paths_path, "r", encoding="utf-8") as f:
            image_paths = json.load(f)

        with open(marker_path, "r", encoding="utf-8") as f:
            marker = json.load(f)

        # Check dimensions, row counts, and saved metadata.
        valid = (
            features.ndim == 2
            and features.shape[1] == manifest["feature_dimension"]
            and len(features) == len(labels) == len(image_paths)
            and len(features) == marker["readable_count"]
            and marker["batch_index"] == i
        )

        if not valid:
            invalid_batches.append(i)
        else:
            total_saved_images += len(features)

    except Exception as exc:
        invalid_batches.append(i)
        print(f"Batch {i} could not be verified: {exc}")

print("Expected batches:", total_batches)
print("Missing batches:", len(missing_batches))
print("Invalid batches:", len(invalid_batches))
print("Verified feature rows:", total_saved_images)

if missing_batches:
    print("Missing batch IDs (first 20):", missing_batches[:20])

if invalid_batches:
    print("Invalid batch IDs (first 20):", invalid_batches[:20])

if not missing_batches and not invalid_batches:
    print("\nSUCCESS: All saved batches passed verification.")
else:
    print("\nWARNING: Repair missing or invalid batches before training.")
```

```python

# ============================================================
# SECTION 13: COMBINE VERIFIED FEATURE BATCHES
# Purpose: Create the combined feature, label, and path files.
# Does not repeat ResNet50 feature extraction.
# ============================================================

import os
import json
import numpy as np

FEATURE_DIR = "/content/drive/MyDrive/CropDiseaseProject/features"
CHECKPOINT_DIR = "/content/drive/MyDrive/CropDiseaseProject/checkpoints"

with open(
    os.path.join(CHECKPOINT_DIR, "feature_manifest.json"),
    "r",
    encoding="utf-8"
) as f:
    manifest = json.load(f)

batch_size = manifest["batch_size"]
total_images = len(manifest["image_paths"])
total_batches = (total_images + batch_size - 1) // batch_size

X_batches = []
y_batches = []
all_paths = []

for i in range(total_batches):
    feature_path = os.path.join(
        FEATURE_DIR, f"features_{i:05d}.npy"
    )
    label_path = os.path.join(
        FEATURE_DIR, f"labels_{i:05d}.npy"
    )
    paths_path = os.path.join(
        FEATURE_DIR, f"paths_{i:05d}.json"
    )

    # Stop if a batch file is missing.
    if not all(os.path.isfile(p) for p in [
        feature_path, label_path, paths_path
    ]):
        raise FileNotFoundError(
            f"Missing files for batch {i}. Check Section 12."
        )

    X_batch = np.load(feature_path)
    y_batch = np.load(label_path)

    with open(paths_path, "r", encoding="utf-8") as f:
        paths_batch = json.load(f)

    assert len(X_batch) == len(y_batch) == len(paths_batch), (
        f"Row mismatch in batch {i}"
    )

    X_batches.append(X_batch)
    y_batches.append(y_batch)
    all_paths.extend(paths_batch)

    if (i + 1) % 100 == 0:
        print(f"Loaded {i + 1}/{total_batches} batches")

# Combine batches in order.
X = np.concatenate(X_batches, axis=0)
y = np.concatenate(y_batches, axis=0)
image_paths = np.asarray(all_paths)

# Validate row alignment.
assert X.shape == (len(y), 2048)
assert len(y) == len(image_paths)
assert len(set(image_paths.tolist())) == len(image_paths)

# Save final arrays to Google Drive.
np.save(os.path.join(FEATURE_DIR, "X_features.npy"), X)
np.save(os.path.join(FEATURE_DIR, "y_labels.npy"), y)
np.save(os.path.join(FEATURE_DIR, "image_paths.npy"), image_paths)

print("\nCombination completed successfully!")
print("Feature shape:", X.shape)
print("Label shape:", y.shape)
print("Image path count:", len(image_paths))
```

```python

# ============================================================
# SECTION 15: PCA DIMENSIONALITY REDUCTION
# Purpose: Reduce 2048 features while retaining 95% variance.
# Important: Fit PCA only on training data to avoid leakage.
# ============================================================

import os
import json
import time
import numpy as np
import joblib
from sklearn.decomposition import PCA

FEATURE_DIR = "/content/drive/MyDrive/CropDiseaseProject/features"
CHECKPOINT_DIR = "/content/drive/MyDrive/CropDiseaseProject/checkpoints"
RESULTS_DIR = "/content/drive/MyDrive/CropDiseaseProject/results"
MODEL_DIR = "/content/drive/MyDrive/CropDiseaseProject/models"

os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(MODEL_DIR, exist_ok=True)

# Load saved features and the reproducible split.
X = np.load(os.path.join(FEATURE_DIR, "X_features.npy"))
y = np.load(os.path.join(FEATURE_DIR, "y_labels.npy"))

split = np.load(
    os.path.join(RESULTS_DIR, "train_test_indices.npz")
)
train_idx = split["train_idx"]
test_idx = split["test_idx"]

# Use float32 to reduce memory usage.
X_train = X[train_idx].astype(np.float32)
X_test = X[test_idx].astype(np.float32)

print("Original training shape:", X_train.shape)
print("Original testing shape:", X_test.shape)

# Fit PCA only on the training data.
pca = PCA(
    n_components=0.95,
    svd_solver="full"
)

start_time = time.time()
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)
pca_time = time.time() - start_time

# Save transformed features and the fitted PCA model.
np.save(
    os.path.join(FEATURE_DIR, "X_train_pca.npy"),
    X_train_pca.astype(np.float32)
)
np.save(
    os.path.join(FEATURE_DIR, "X_test_pca.npy"),
    X_test_pca.astype(np.float32)
)

joblib.dump(
    pca,
    os.path.join(MODEL_DIR, "resnet50_pca.joblib")
)

# Save PCA information for the final report.
pca_info = {
    "original_features": int(X_train.shape[1]),
    "reduced_features": int(X_train_pca.shape[1]),
    "variance_retained": float(
        np.sum(pca.explained_variance_ratio_)
    ),
    "fit_transform_time_seconds": round(pca_time, 2)
}

with open(
    os.path.join(RESULTS_DIR, "pca_summary.json"),
    "w",
    encoding="utf-8"
) as f:
    json.dump(pca_info, f, indent=4)

print("\nPCA completed successfully.")
print("Reduced training shape:", X_train_pca.shape)
print("Reduced testing shape:", X_test_pca.shape)
print(
    "Variance retained:",
    round(pca_info["variance_retained"] * 100, 2), "%"
)
print("PCA time:", round(pca_time, 2), "seconds")
print("PCA model and transformed features saved to Drive.")
```

```python

# ============================================================
# SECTION 22: EFFICIENTNETB0 FEATURE EXTRACTION
# Features are saved in resumable batches on Google Drive.
# ============================================================

import os
import json
import numpy as np
import tensorflow as tf

from PIL import Image, ImageFile
from tensorflow.keras.applications import EfficientNetB0

# ------------------------------------------------------------
# 1. Paths and settings
# ------------------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/CropDiseaseProject"
DATASET_DIR = "/content/drive/MyDrive/crop_disease_dataset"

FEATURE_DIR = os.path.join(
    BASE_DIR, "features", "efficientnetb0"
)
CHECKPOINT_DIR = os.path.join(
    BASE_DIR, "checkpoints", "efficientnetb0"
)

os.makedirs(FEATURE_DIR, exist_ok=True)
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

IMAGE_PATHS_FILE = os.path.join(
    BASE_DIR, "features", "image_paths.npy"
)
LABELS_FILE = os.path.join(
    BASE_DIR, "features", "y_labels.npy"
)

BATCH_SIZE = 32
IMAGE_SIZE = (224, 224)
FEATURE_DIM = 1280

# ------------------------------------------------------------
# 2. Load image paths and labels
# ------------------------------------------------------------

image_paths = np.load(IMAGE_PATHS_FILE, allow_pickle=True)
labels = np.load(LABELS_FILE)

assert len(image_paths) == len(labels), (
    "Image paths and labels have different lengths."
)

print("Images to process:", len(image_paths))
print("Batch size:", BATCH_SIZE)

# ------------------------------------------------------------
# 3. Load the pretrained feature extractor
# ------------------------------------------------------------

model = EfficientNetB0(
    weights="imagenet",
    include_top=False,
    pooling="avg",
    input_shape=(224, 224, 3)
)
model.trainable = False

print("EfficientNetB0 ready.")

# ------------------------------------------------------------
# 4. Resume from the last saved batch
# ------------------------------------------------------------

MANIFEST_PATH = os.path.join(
    CHECKPOINT_DIR, "extraction_manifest.json"
)

# Each completed batch contains its features, labels,
# original indices, and image paths.
if os.path.exists(MANIFEST_PATH):
    with open(MANIFEST_PATH, "r", encoding="utf-8") as f:
        manifest = json.load(f)
else:
    manifest = {
        "completed_batches": [],
        "failed_indices": [],
        "feature_dimension": FEATURE_DIM,
        "batch_size": BATCH_SIZE
    }

completed_batches = set(manifest["completed_batches"])
total_batches = (len(image_paths) + BATCH_SIZE - 1) // BATCH_SIZE

print("Total batches:", total_batches)
print("Already completed:", len(completed_batches))

# ------------------------------------------------------------
# 5. Extract and save each batch
# ------------------------------------------------------------

for batch_id in range(total_batches):

    if batch_id in completed_batches:
        continue

    start_idx = batch_id * BATCH_SIZE
    end_idx = min(start_idx + BATCH_SIZE, len(image_paths))

    batch_features = []
    batch_labels = []
    batch_paths = []
    batch_indices = []
    failed_indices = []

    for idx in range(start_idx, end_idx):
        try:
            # Load and resize the image.
            with Image.open(str(image_paths[idx])) as img:
                img = img.convert("RGB")
                img = img.resize(IMAGE_SIZE)
                image_array = np.asarray(img, dtype=np.float32)

            batch_features.append(image_array)
            batch_labels.append(labels[idx])
            batch_paths.append(str(image_paths[idx]))
            batch_indices.append(idx)

        except Exception as exc:
            failed_indices.append(idx)

            # Record the failure so it can be reviewed later.
            with open(
                os.path.join(CHECKPOINT_DIR, "failed_images.log"),
                "a",
                encoding="utf-8"
            ) as log:
                log.write(f"{idx}\t{image_paths[idx]}\t{exc}\n")

    # Save even if every image in a batch is unreadable.
    # Empty batches are represented by correctly shaped arrays.
    if batch_features:
        batch_array = np.stack(batch_features, axis=0)

        # EfficientNetB0 expects pixel values in [0, 255]
        # for its built-in preprocessing.
        feature_array = model.predict(
            batch_array,
            batch_size=BATCH_SIZE,
            verbose=0
        ).astype(np.float32)
    else:
        feature_array = np.empty(
            (0, FEATURE_DIM), dtype=np.float32
        )

    # Persist this batch to Drive.
    batch_file = os.path.join(
        FEATURE_DIR, f"batch_{batch_id:04d}.npz"
    )

    temp_file = batch_file + ".tmp.npz"

    np.savez_compressed(
        temp_file,
        features=feature_array,
        labels=np.asarray(batch_labels, dtype=labels.dtype),
        paths=np.asarray(batch_paths, dtype=str),
        indices=np.asarray(batch_indices, dtype=np.int64)
    )

    os.replace(temp_file, batch_file)

    # Update the manifest only after the batch is saved.
    completed_batches.add(batch_id)
    manifest["completed_batches"] = sorted(completed_batches)
    manifest["failed_indices"] = sorted(
        set(manifest["failed_indices"] + failed_indices)
    )

    temp_manifest = MANIFEST_PATH + ".tmp"
    with open(temp_manifest, "w", encoding="utf-8") as f:
        json.dump(manifest, f, indent=2)
    os.replace(temp_manifest, MANIFEST_PATH)

    if (
        (batch_id + 1) % 25 == 0
        or batch_id == total_batches - 1
    ):
        print(
            f"Completed batch {batch_id + 1}/{total_batches} | "
            f"Saved images: {sum(1 for _ in completed_batches)} batches"
        )

print("\nFeature extraction pass completed.")
print("Completed batches:", len(completed_batches))
print("Failed image indices:", len(manifest["failed_indices"]))
print("Features saved in:", FEATURE_DIR)
```

```python

# ============================================================
# SECTION 24: COMBINE RESNET50 + EFFICIENTNETB0 FEATURES
# ============================================================

import os
import glob
import numpy as np

BASE_DIR = "/content/drive/MyDrive/CropDiseaseProject"

RESNET_DIR = os.path.join(BASE_DIR, "features")
EFFICIENTNET_DIR = os.path.join(
    BASE_DIR, "features", "efficientnetb0"
)
HYBRID_DIR = os.path.join(BASE_DIR, "features", "hybrid")

os.makedirs(HYBRID_DIR, exist_ok=True)

# ------------------------------------------------------------
# 1. Load existing ResNet50 features and labels
# ------------------------------------------------------------

X_resnet = np.load(
    os.path.join(RESNET_DIR, "X_features.npy")
)
y_resnet = np.load(
    os.path.join(RESNET_DIR, "y_labels.npy")
)
resnet_paths = np.load(
    os.path.join(RESNET_DIR, "image_paths.npy"),
    allow_pickle=True
)

print("ResNet50 shape:", X_resnet.shape)
print("ResNet50 labels:", y_resnet.shape)

# ------------------------------------------------------------
# 2. Load EfficientNetB0 batches and restore dataset order
# ------------------------------------------------------------

batch_files = sorted(
    glob.glob(os.path.join(EFFICIENTNET_DIR, "batch_*.npz"))
)

efficient_features = np.empty(
    (len(resnet_paths), 1280), dtype=np.float32
)
efficient_labels = np.empty(
    len(resnet_paths), dtype=y_resnet.dtype
)
efficient_paths = np.empty(
    len(resnet_paths), dtype=object
)
seen = np.zeros(len(resnet_paths), dtype=bool)

for batch_file in batch_files:
    with np.load(batch_file, allow_pickle=False) as data:
        indices = data["indices"]
        features = data["features"]
        labels_batch = data["labels"]
        paths_batch = data["paths"]

        efficient_features[indices] = features
        efficient_labels[indices] = labels_batch
        efficient_paths[indices] = paths_batch
        seen[indices] = True

# ------------------------------------------------------------
# 3. Verify that both representations align
# ------------------------------------------------------------

assert seen.all(), "Some EfficientNetB0 indices are missing."
assert len(resnet_paths) == len(efficient_features)
assert np.array_equal(y_resnet, efficient_labels), (
    "Labels do not match in dataset order."
)
assert np.array_equal(
    resnet_paths.astype(str), efficient_paths.astype(str)
), "Image paths do not match in dataset order."

print("Feature alignment verified.")

# ------------------------------------------------------------
# 4. Concatenate the feature vectors
# ------------------------------------------------------------

X_hybrid = np.concatenate(
    [X_resnet.astype(np.float32), efficient_features],
    axis=1
)

y_hybrid = y_resnet.copy()

assert X_hybrid.shape == (len(y_hybrid), 3328)

# ------------------------------------------------------------
# 5. Save hybrid features to Google Drive
# ------------------------------------------------------------

np.save(
    os.path.join(HYBRID_DIR, "X_hybrid_features.npy"),
    X_hybrid
)
np.save(
    os.path.join(HYBRID_DIR, "y_hybrid_labels.npy"),
    y_hybrid
)
np.save(
    os.path.join(HYBRID_DIR, "image_paths.npy"),
    resnet_paths
)

print("\nHybrid feature combination completed!")
print("Hybrid feature shape:", X_hybrid.shape)
print("Hybrid label shape:", y_hybrid.shape)
print("Saved in:", HYBRID_DIR)
```

```python

# ============================================================
# SECTION 27: PCA ON HYBRID FEATURES
# Retain 95% of the training data variance
# ============================================================

import os
import numpy as np
import joblib

from sklearn.decomposition import PCA

BASE_DIR = "/content/drive/MyDrive/CropDiseaseProject"
FEATURE_DIR = os.path.join(BASE_DIR, "features", "hybrid")
MODEL_DIR = os.path.join(BASE_DIR, "models")
RESULTS_DIR = os.path.join(BASE_DIR, "results")

os.makedirs(FEATURE_DIR, exist_ok=True)
os.makedirs(MODEL_DIR, exist_ok=True)
os.makedirs(RESULTS_DIR, exist_ok=True)

# Fit PCA on training data ONLY.
pca = PCA(
    n_components=0.95,
    svd_solver="full"
)

X_train_pca = pca.fit_transform(X_train)

# Apply the same PCA transformation to testing data.
X_test_pca = pca.transform(X_test)

# Save the fitted PCA model for future predictions.
pca_path = os.path.join(MODEL_DIR, "PCA_Hybrid_95.joblib")
joblib.dump(pca, pca_path)

# Save transformed features permanently in Google Drive.
np.save(
    os.path.join(FEATURE_DIR, "X_train_pca_95.npy"),
    X_train_pca
)
np.save(
    os.path.join(FEATURE_DIR, "X_test_pca_95.npy"),
    X_test_pca
)

# Print dimensionality reduction statistics.
original_dimensions = X_train.shape[1]
reduced_dimensions = X_train_pca.shape[1]

print("Original feature dimensions:", original_dimensions)
print("Reduced feature dimensions:", reduced_dimensions)
print("Original training shape:", X_train.shape)
print("Reduced training shape:", X_train_pca.shape)
print("Original testing shape:", X_test.shape)
print("Reduced testing shape:", X_test_pca.shape)
print("Explained variance retained:",
      round(np.sum(pca.explained_variance_ratio_) * 100, 2), "%")
print("Dimensionality reduction:",
      round((1 - reduced_dimensions / original_dimensions) * 100, 2), "%")
print("PCA model saved to:", pca_path)
print("PCA feature files saved successfully.")
```

```python
# ============================================================
# CREATE IMAGE PATHS AND LABELS
# Purpose: Collect image paths and class labels from the dataset
# ============================================================

import os
import numpy as np
from PIL import Image
from tqdm.auto import tqdm

# Dataset directory
DATASET_DIR = "/content/drive/MyDrive/crop_disease_dataset_unzip"

# Save location in Google Drive
FEATURE_DIR = "/content/drive/MyDrive/CropDiseaseProject/features/vgg16"
os.makedirs(FEATURE_DIR, exist_ok=True)

IMAGE_PATHS_FILE = os.path.join(FEATURE_DIR, "image_paths.npy")
LABELS_FILE = os.path.join(FEATURE_DIR, "labels.npy")

# Supported image formats
IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

# Collect class folders
class_names = sorted([
    folder for folder in os.listdir(DATASET_DIR)
    if os.path.isdir(os.path.join(DATASET_DIR, folder))
])

print("Number of classes:", len(class_names))
print("Classes:", class_names)

# Create class-to-label mapping
class_to_label = {
    class_name: index
    for index, class_name in enumerate(class_names)
}

# Collect image paths and labels
image_paths = []
labels = []
failed_images = []

for class_name in tqdm(class_names, desc="Processing classes"):
    class_folder = os.path.join(DATASET_DIR, class_name)

    for root, dirs, files in os.walk(class_folder):
        for file in files:
            if file.lower().endswith(IMAGE_EXTENSIONS):
                image_path = os.path.join(root, file)

                # Verify that the image can be opened
                try:
                    with Image.open(image_path) as img:
                        img.verify()

                    image_paths.append(image_path)
                    labels.append(class_to_label[class_name])

                except Exception:
                    failed_images.append(image_path)

# Convert to NumPy arrays
image_paths = np.array(image_paths)
labels = np.array(labels, dtype=np.int32)

# Save paths and labels to Drive
np.save(IMAGE_PATHS_FILE, image_paths)
np.save(LABELS_FILE, labels)

# Print summary
print("\n========== DATASET SUMMARY ==========")
print("Dataset path:", DATASET_DIR)
print("Number of classes:", len(class_names))
print("Valid images:", len(image_paths))
print("Labels:", len(labels))
print("Unreadable images skipped:", len(failed_images))
print("Image paths saved to:", IMAGE_PATHS_FILE)
print("Labels saved to:", LABELS_FILE)
print("====================================")
```

```python
# ============================================================
#  VGG16 FEATURE EXTRACTION
# ============================================================

import os
import numpy as np
import tensorflow as tf
from tensorflow.keras.applications import VGG16
from tensorflow.keras.applications.vgg16 import preprocess_input
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tqdm.auto import tqdm

# ----------------------------
# 1. Paths
# ----------------------------
PROJECT_DIR = "/content/drive/MyDrive/CropDiseaseProject"
FEATURE_DIR = os.path.join(PROJECT_DIR, "features", "vgg16")
CHECKPOINT_DIR = os.path.join(PROJECT_DIR, "checkpoints", "vgg16")

os.makedirs(FEATURE_DIR, exist_ok=True)
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

IMAGE_PATHS_FILE = os.path.join(FEATURE_DIR, "image_paths.npy")
LABELS_FILE = os.path.join(FEATURE_DIR, "labels.npy")

FEATURES_FILE = os.path.join(FEATURE_DIR, "vgg16_features.npy")
FEATURE_LABELS_FILE = os.path.join(FEATURE_DIR, "vgg16_labels.npy")
CHECKPOINT_FILE = os.path.join(CHECKPOINT_DIR, "vgg16_checkpoint.npz")

# ----------------------------
# 2. Load saved paths and labels
# ----------------------------
image_paths = np.load(IMAGE_PATHS_FILE, allow_pickle=True)
labels = np.load(LABELS_FILE)

print("Images:", len(image_paths))
print("Labels:", len(labels))

# ----------------------------
# 3. Load pretrained VGG16
# ----------------------------
base_model = VGG16(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

# Global average pooling gives 512 features per image
model = tf.keras.Sequential([
    base_model,
    tf.keras.layers.GlobalAveragePooling2D()
])

model.trainable = False

print("VGG16 feature extractor loaded.")
print("Feature dimension:", model.output_shape[-1])

# ----------------------------
# 4. Resume from checkpoint if available
# ----------------------------
BATCH_SIZE = 32
SAVE_EVERY = 100  # save progress every 100 batches

if os.path.exists(CHECKPOINT_FILE):
    checkpoint = np.load(CHECKPOINT_FILE, allow_pickle=True)
    start_index = int(checkpoint["next_index"])
    saved_features = checkpoint["features"].tolist()
    saved_indices = checkpoint["indices"].tolist()
    print("Resuming from image index:", start_index)
else:
    start_index = 0
    saved_features = []
    saved_indices = []
    print("Starting feature extraction from the beginning.")

# ----------------------------
# 5. Extract features in batches
# ----------------------------
for start in tqdm(range(start_index, len(image_paths), BATCH_SIZE),
                  desc="VGG16 feature extraction"):

    batch_paths = image_paths[start:start + BATCH_SIZE]
    batch_images = []
    batch_indices = []

    for offset, image_path in enumerate(batch_paths):
        try:
            img = load_img(image_path, target_size=(224, 224))
            img_array = img_to_array(img)
            batch_images.append(img_array)
            batch_indices.append(start + offset)
        except Exception as e:
            print("Skipping unreadable image:", image_path)

    if not batch_images:
        continue

    batch_array = np.array(batch_images, dtype=np.float32)
    batch_array = preprocess_input(batch_array)

    batch_features = model.predict(batch_array, verbose=0)

    saved_features.extend(batch_features)
    saved_indices.extend(batch_indices)

    # Periodic checkpoint
    if (start // BATCH_SIZE + 1) % SAVE_EVERY == 0:
        np.savez(
            CHECKPOINT_FILE,
            next_index=start + BATCH_SIZE,
            features=np.array(saved_features, dtype=np.float32),
            indices=np.array(saved_indices, dtype=np.int32)
        )
        print(f"Checkpoint saved at image index {start + BATCH_SIZE}")

# ----------------------------
# 6. Save final features
# ----------------------------
features = np.array(saved_features, dtype=np.float32)
feature_indices = np.array(saved_indices, dtype=np.int32)

# Align labels with successfully extracted image indices
extracted_labels = labels[feature_indices]

np.save(FEATURES_FILE, features)
np.save(FEATURE_LABELS_FILE, extracted_labels)

# Save final checkpoint
np.savez(
    CHECKPOINT_FILE,
    next_index=len(image_paths),
    features=features,
    indices=feature_indices
)

print("\n========== VGG16 EXTRACTION COMPLETE ==========")
print("Feature shape:", features.shape)
print("Label shape:", extracted_labels.shape)
print("Features saved to:", FEATURES_FILE)
print("Labels saved to:", FEATURE_LABELS_FILE)
print("================================================")
```

```python
#  VGG16 - Train/Test Split and PCA

import os
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Paths
PROJECT_DIR = "/content/drive/MyDrive/CropDiseaseProject"
VGG_DIR = os.path.join(PROJECT_DIR, "features", "vgg16")
RESULTS_DIR = os.path.join(PROJECT_DIR, "results", "vgg16")
os.makedirs(RESULTS_DIR, exist_ok=True)

# Load extracted VGG16 features and labels
X = np.load(os.path.join(VGG_DIR, "vgg16_features.npy"))
y = np.load(os.path.join(VGG_DIR, "vgg16_labels.npy"))

print("VGG16 features:", X.shape)
print("Labels:", y.shape)

# Stratified train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)

# Standardize features (fit only on training data)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# PCA retaining 95% variance
pca = PCA(n_components=0.95, random_state=42)
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)

print("PCA training shape:", X_train_pca.shape)
print("PCA testing shape:", X_test_pca.shape)
print("Number of PCA components:", pca.n_components_)
print("Variance retained:", round(np.sum(pca.explained_variance_ratio_), 4))

# Save split and transformed data
np.savez_compressed(
    os.path.join(RESULTS_DIR, "vgg16_train_test_pca.npz"),
    X_train=X_train,
    X_test=X_test,
    y_train=y_train,
    y_test=y_test,
    X_train_scaled=X_train_scaled,
    X_test_scaled=X_test_scaled,
    X_train_pca=X_train_pca,
    X_test_pca=X_test_pca
)

import joblib
joblib.dump(scaler, os.path.join(RESULTS_DIR, "vgg16_scaler.joblib"))
joblib.dump(pca, os.path.join(RESULTS_DIR, "vgg16_pca.joblib"))

print("Saved VGG16 split, scaler, and PCA outputs to:", RESULTS_DIR)
```

```python
#  VGG16 - Random Forest (With and Without PCA)

import os
import time
import joblib
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

RESULTS_DIR = "/content/drive/MyDrive/CropDiseaseProject/results/vgg16"
data = np.load(os.path.join(RESULTS_DIR, "vgg16_train_test_pca.npz"))

X_train = data["X_train"]
X_test = data["X_test"]
X_train_pca = data["X_train_pca"]
X_test_pca = data["X_test_pca"]
y_train = data["y_train"]
y_test = data["y_test"]

rf_results = []

for name, Xtr, Xte in [
    ("Without_PCA", X_train, X_test),
    ("With_PCA", X_train_pca, X_test_pca)
]:
    print(f"\nTraining Random Forest - {name}")
    start = time.time()

    model = RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )
    model.fit(Xtr, y_train)
    y_pred = model.predict(Xte)

    metrics = {
        "Model": "Random Forest",
        "Feature_Set": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision_Weighted": precision_score(y_test, y_pred, average="weighted", zero_division=0),
        "Recall_Weighted": recall_score(y_test, y_pred, average="weighted", zero_division=0),
        "F1_Weighted": f1_score(y_test, y_pred, average="weighted", zero_division=0),
        "Training_Time_Seconds": round(time.time() - start, 2)
    }
    rf_results.append(metrics)

    joblib.dump(
        model,
        os.path.join(RESULTS_DIR, f"vgg16_random_forest_{name.lower()}.joblib")
    )

    print(metrics)

import pandas as pd
rf_df = pd.DataFrame(rf_results)
rf_df.to_csv(os.path.join(RESULTS_DIR, "vgg16_random_forest_results.csv"), index=False)
display(rf_df)
```

```python
# ============================================================
# SECTION 37: LOAD AND ALIGN VGG16, RESNET50, EFFICIENTNETB0 FEATURES
# Purpose: Put all three feature sets in the same image order and
# reuse your saved 80:20 split (train_test_indices.npz).
# ============================================================

import os, glob, json, time, joblib
import numpy as np

BASE_DIR    = "/content/drive/MyDrive/CropDiseaseProject"
FEATURE_DIR = os.path.join(BASE_DIR, "features")
RESULTS_DIR = os.path.join(BASE_DIR, "results")
MODEL_DIR   = os.path.join(BASE_DIR, "models")
CKPT_DIR    = os.path.join(BASE_DIR, "checkpoints")
for d in [RESULTS_DIR, MODEL_DIR]:
    os.makedirs(d, exist_ok=True)

def img_key(path):
    """class_folder/filename: works even if the dataset folder was renamed."""
    return "/".join(str(path).replace("\\", "/").split("/")[-2:])

# ---------- ResNet50 (master order) ----------
X_res         = np.load(os.path.join(FEATURE_DIR, "X_features.npy"), mmap_mode="r")
y_master      = np.load(os.path.join(FEATURE_DIR, "y_labels.npy"))
master_paths  = np.load(os.path.join(FEATURE_DIR, "image_paths.npy"), allow_pickle=True)
master_keys   = np.array([img_key(p) for p in master_paths])
N = len(master_keys)
print("ResNet50 :", X_res.shape)
print("Duplicate image keys in master list:", N - len(set(master_keys)))

# ---------- EfficientNetB0 (batch files -> master order) ----------
X_eff   = np.zeros((N, 1280), dtype=np.float32)
has_eff = np.zeros(N, dtype=bool)
for f in sorted(glob.glob(os.path.join(FEATURE_DIR, "efficientnetb0", "batch_*.npz"))):
    with np.load(f, allow_pickle=False) as d:
        X_eff[d["indices"]]   = d["features"]
        has_eff[d["indices"]] = True
print("EfficientNetB0:", X_eff.shape, "| missing:", int((~has_eff).sum()))

# ---------- VGG16 (match by class/filename) ----------
vgg_dir   = os.path.join(FEATURE_DIR, "vgg16")
vgg_feats = np.load(os.path.join(vgg_dir, "vgg16_features.npy"))
vgg_lbls  = np.load(os.path.join(vgg_dir, "vgg16_labels.npy"))
vgg_paths = np.load(os.path.join(vgg_dir, "image_paths.npy"), allow_pickle=True)

# Rows in vgg16_features.npy correspond to the images that were readable,
# whose positions are stored in the checkpoint file.
ck = np.load(os.path.join(CKPT_DIR, "vgg16", "vgg16_checkpoint.npz"), allow_pickle=True)
vgg_idx = ck["indices"]
assert len(vgg_idx) == len(vgg_feats), "VGG16 checkpoint indices do not match the feature rows."
vgg_keys = np.array([img_key(p) for p in vgg_paths[vgg_idx]])

key_to_row = {k: i for i, k in enumerate(master_keys)}
X_vgg   = np.zeros((N, 512), dtype=np.float32)
has_vgg = np.zeros(N, dtype=bool)
label_mismatch = 0
for r, k in enumerate(vgg_keys):
    m = key_to_row.get(k)
    if m is None:
        continue
    X_vgg[m] = vgg_feats[r]
    has_vgg[m] = True
    # VGG16 labels were built from its own folder listing, so compare by class name instead of number
    label_mismatch += int(k.split("/")[0] != master_keys[m].split("/")[0])
print("VGG16    :", X_vgg.shape, "| matched:", int(has_vgg.sum()), "| missing:", int((~has_vgg).sum()))
print("Class-folder mismatches:", label_mismatch)

# ---------- Common images and the saved split ----------
common = has_eff & has_vgg
with np.load(os.path.join(RESULTS_DIR, "train_test_indices.npz")) as s:
    train_idx = s["train_idx"][common[s["train_idx"]]]
    test_idx  = s["test_idx"][common[s["test_idx"]]]

y_train, y_test = y_master[train_idx], y_master[test_idx]
with open(os.path.join(CKPT_DIR, "feature_manifest.json"), encoding="utf-8") as f:
    class_names = json.load(f)["class_names"]

print("\nTrain:", len(train_idx), "| Test:", len(test_idx), "| Classes:", len(class_names))

# Feature sets (hybrids are concatenated later, only for the rows needed)
FEATURE_SETS = {
    "VGG16":              [X_vgg],
    "ResNet50":           [X_res],
    "EfficientNetB0":     [X_eff],
    "Hybrid_Res_Eff":     [X_res, X_eff],
    "Hybrid_VGG_Res_Eff": [X_vgg, X_res, X_eff],
}
```

```python
# ============================================================
# SECTION 37: LOAD AND ALIGN VGG16, RESNET50 AND EFFICIENTNETB0
# Purpose: Put all three feature sets in the same image order
# and reuse the saved 80:20 train-test split.
# ============================================================

# ------------------------------------------------------------
# 0. Mount Google Drive and stop if it is not really mounted
# (prevents creating fake local folders that block the mount)
# ------------------------------------------------------------

import os
from google.colab import drive

drive.mount("/content/drive")
assert os.path.ismount("/content/drive"), (
    "Google Drive is NOT mounted. Do not continue."
)

import os
import glob
import json
import time
import joblib
import numpy as np

# ------------------------------------------------------------
# 1. Define project paths
# ------------------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/CropDiseaseProject"
FEATURE_DIR = os.path.join(BASE_DIR, "features")
CHECKPOINT_DIR = os.path.join(BASE_DIR, "checkpoints")
MODEL_DIR = os.path.join(BASE_DIR, "models")
RESULTS_DIR = os.path.join(BASE_DIR, "results")

os.makedirs(MODEL_DIR, exist_ok=True)
os.makedirs(RESULTS_DIR, exist_ok=True)

# ------------------------------------------------------------
# 2. Helper: build a path-independent image key
# The VGG16 run used a different dataset folder, so images are
# matched by "class_folder/filename" instead of the full path.
# ------------------------------------------------------------

def image_key(path):
    parts = str(path).replace("\\", "/").split("/")
    return "/".join(parts[-2:])

# ------------------------------------------------------------
# 3. Load ResNet50 features (this order is the master order)
# ------------------------------------------------------------

X_resnet = np.load(
    os.path.join(FEATURE_DIR, "X_features.npy"),
    mmap_mode="r"
)
y_master = np.load(
    os.path.join(FEATURE_DIR, "y_labels.npy")
)
master_paths = np.load(
    os.path.join(FEATURE_DIR, "image_paths.npy"),
    allow_pickle=True
)

master_keys = np.array([image_key(p) for p in master_paths])
total_images = len(master_keys)

print("ResNet50 shape:", X_resnet.shape)
print("Duplicate image keys:", total_images - len(set(master_keys)))

# ------------------------------------------------------------
# 4. Load EfficientNetB0 batches and restore the master order
# ------------------------------------------------------------

X_efficient = np.zeros((total_images, 1280), dtype=np.float32)
has_efficient = np.zeros(total_images, dtype=bool)

batch_files = sorted(
    glob.glob(
        os.path.join(FEATURE_DIR, "efficientnetb0", "batch_*.npz")
    )
)

for batch_file in batch_files:
    with np.load(batch_file, allow_pickle=False) as data:
        X_efficient[data["indices"]] = data["features"]
        has_efficient[data["indices"]] = True

print("EfficientNetB0 shape:", X_efficient.shape)
print("EfficientNetB0 missing images:", int((~has_efficient).sum()))

# ------------------------------------------------------------
# 5. Load VGG16 features and match them to the master order
# ------------------------------------------------------------

vgg_dir = os.path.join(FEATURE_DIR, "vgg16")

vgg_features = np.load(os.path.join(vgg_dir, "vgg16_features.npy"))
vgg_paths = np.load(
    os.path.join(vgg_dir, "image_paths.npy"),
    allow_pickle=True
)

# The checkpoint stores which images produced each feature row.
vgg_checkpoint = np.load(
    os.path.join(CHECKPOINT_DIR, "vgg16", "vgg16_checkpoint.npz"),
    allow_pickle=True
)
vgg_indices = vgg_checkpoint["indices"]

assert len(vgg_indices) == len(vgg_features), (
    "VGG16 checkpoint indices do not match the feature rows."
)

vgg_keys = np.array([image_key(p) for p in vgg_paths[vgg_indices]])
key_to_master_row = {key: row for row, key in enumerate(master_keys)}

X_vgg = np.zeros((total_images, 512), dtype=np.float32)
has_vgg = np.zeros(total_images, dtype=bool)
class_mismatch = 0

for row, key in enumerate(vgg_keys):
    master_row = key_to_master_row.get(key)

    # Skip images that do not exist in the ResNet50 list.
    if master_row is None:
        continue

    X_vgg[master_row] = vgg_features[row]
    has_vgg[master_row] = True

    # Confirm the class folder is the same in both lists.
    if key.split("/")[0] != master_keys[master_row].split("/")[0]:
        class_mismatch += 1

print("VGG16 shape:", X_vgg.shape)
print("VGG16 matched images:", int(has_vgg.sum()))
print("VGG16 missing images:", int((~has_vgg).sum()))
print("Class-folder mismatches:", class_mismatch)

# ------------------------------------------------------------
# 6. Reuse the saved train-test split
# Only images that have all three feature sets are kept.
# ------------------------------------------------------------

has_all_features = has_efficient & has_vgg

with np.load(
    os.path.join(RESULTS_DIR, "train_test_indices.npz")
) as split:
    train_idx = split["train_idx"][has_all_features[split["train_idx"]]]
    test_idx = split["test_idx"][has_all_features[split["test_idx"]]]

y_train = y_master[train_idx]
y_test = y_master[test_idx]

# Load class names for the classification reports.
with open(
    os.path.join(CHECKPOINT_DIR, "feature_manifest.json"),
    "r",
    encoding="utf-8"
) as f:
    class_names = json.load(f)["class_names"]

# ------------------------------------------------------------
# 7. Define the feature sets to evaluate
# Hybrid sets are concatenated later, only for the rows needed.
# ------------------------------------------------------------

FEATURE_SETS = {
    "VGG16": [X_vgg],
    "ResNet50": [X_resnet],
    "EfficientNetB0": [X_efficient],
    "Hybrid_Res_Eff": [X_resnet, X_efficient],
    "Hybrid_VGG_Res_Eff": [X_vgg, X_resnet, X_efficient],
}

print("\nTraining samples:", len(train_idx))
print("Testing samples:", len(test_idx))
print("Number of classes:", len(class_names))
print("Feature sets ready:", list(FEATURE_SETS.keys()))
```

```python
# ============================================================
# SECTION 38: STANDARD SCALING + PCA FOR EVERY FEATURE SET
# Retain 95% of the training data variance
# Important: Fit the scaler and PCA on training data ONLY.
# ============================================================

import os
import json
import time
import joblib
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

PCA_DIR = os.path.join(FEATURE_DIR, "pca_all")
os.makedirs(PCA_DIR, exist_ok=True)

# Stores the reduced training and testing features per feature set.
pca_data = {}
pca_summary = {}

for feature_name, parts in FEATURE_SETS.items():

    print("=" * 55)
    print("Preparing PCA for:", feature_name)

    # --------------------------------------------------------
    # 1. Select the training and testing rows
    # Hybrid sets are joined side by side (concatenation).
    # --------------------------------------------------------

    X_train = np.concatenate(
        [np.asarray(p[train_idx], dtype=np.float32) for p in parts],
        axis=1
    )
    X_test = np.concatenate(
        [np.asarray(p[test_idx], dtype=np.float32) for p in parts],
        axis=1
    )

    # --------------------------------------------------------
    # 2. Standardize the features
    # SVM and KNN depend on distances, so every feature must
    # be on a similar scale before PCA is applied.
    # --------------------------------------------------------

    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # --------------------------------------------------------
    # 3. Fit PCA on training data ONLY
    # --------------------------------------------------------

    pca = PCA(
        n_components=0.95,
        svd_solver="full",
        random_state=42
    )

    start_time = time.time()
    X_train_pca = pca.fit_transform(X_train_scaled)

    # Apply the same transformation to the testing data.
    X_test_pca = pca.transform(X_test_scaled)
    pca_time = time.time() - start_time

    # --------------------------------------------------------
    # 4. Save the scaler, PCA model and transformed features
    # --------------------------------------------------------

    joblib.dump(
        scaler,
        os.path.join(MODEL_DIR, f"{feature_name}_scaler.joblib")
    )
    joblib.dump(
        pca,
        os.path.join(MODEL_DIR, f"{feature_name}_PCA_95.joblib")
    )
    np.save(
        os.path.join(PCA_DIR, f"{feature_name}_X_train_pca.npy"),
        X_train_pca.astype(np.float32)
    )
    np.save(
        os.path.join(PCA_DIR, f"{feature_name}_X_test_pca.npy"),
        X_test_pca.astype(np.float32)
    )

    pca_data[feature_name] = (
        X_train_pca.astype(np.float32),
        X_test_pca.astype(np.float32)
    )

    # --------------------------------------------------------
    # 5. Record the dimensionality reduction statistics
    # --------------------------------------------------------

    pca_summary[feature_name] = {
        "original_dimensions": int(X_train.shape[1]),
        "dimensions_after_pca": int(X_train_pca.shape[1]),
        "variance_retained_percent": round(
            float(np.sum(pca.explained_variance_ratio_)) * 100, 2
        ),
        "pca_time_seconds": round(pca_time, 2)
    }

    print("Original dimensions:", X_train.shape[1])
    print("Reduced dimensions:", X_train_pca.shape[1])
    print(
        "Variance retained:",
        pca_summary[feature_name]["variance_retained_percent"], "%"
    )
    print("PCA time:", round(pca_time, 2), "seconds")

# Save the PCA summary for the final report.
with open(
    os.path.join(RESULTS_DIR, "pca_summary_all_features.json"),
    "w",
    encoding="utf-8"
) as f:
    json.dump(pca_summary, f, indent=4)

print("\nPCA completed for all feature sets.")
```

```python
# ============================================================
# SECTION 39: SVM WITH PCA ON ALL FEATURE SETS
# Model: Support Vector Machine (RBF kernel)
# ============================================================

import os
import json
import time
import joblib
import numpy as np

from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

svm_results = {}

for feature_name, (X_train_pca, X_test_pca) in pca_data.items():

    print("=" * 55)
    print("Training SVM on", feature_name, "features WITH PCA")

    # Confirm that the PCA features and labels are aligned.
    assert X_train_pca.shape[0] == len(y_train)
    assert X_test_pca.shape[0] == len(y_test)

    # --------------------------------------------------------
    # 1. Define the SVM classifier
    # C controls the penalty for misclassification.
    # gamma="scale" sets the RBF width from the data variance.
    # --------------------------------------------------------

    svm_pca = SVC(
        kernel="rbf",
        C=10,
        gamma="scale",
        random_state=42
    )

    # --------------------------------------------------------
    # 2. Train the model and measure the training time
    # --------------------------------------------------------

    start_time = time.time()
    svm_pca.fit(X_train_pca, y_train)
    svm_train_time = time.time() - start_time

    # --------------------------------------------------------
    # 3. Predict the test set and measure the prediction time
    # --------------------------------------------------------

    start_time = time.time()
    svm_pred = svm_pca.predict(X_test_pca)
    svm_predict_time = time.time() - start_time

    # --------------------------------------------------------
    # 4. Calculate the evaluation metrics
    # --------------------------------------------------------

    run_name = f"SVM_{feature_name}_PCA"

    svm_results[run_name] = {
        "feature_extraction": feature_name,
        "classifier": "SVM",
        "dimensions_after_pca": int(X_train_pca.shape[1]),
        "variance_retained_percent": 95.0,
        "accuracy": float(accuracy_score(y_test, svm_pred)),
        "macro_precision": float(
            precision_score(y_test, svm_pred, average="macro",
                            zero_division=0)
        ),
        "macro_recall": float(
            recall_score(y_test, svm_pred, average="macro",
                         zero_division=0)
        ),
        "macro_f1": float(
            f1_score(y_test, svm_pred, average="macro",
                     zero_division=0)
        ),
        "training_time_seconds": float(svm_train_time),
        "prediction_time_seconds": float(svm_predict_time)
    }

    # --------------------------------------------------------
    # 5. Save the model, report and confusion matrix
    # --------------------------------------------------------

    joblib.dump(
        svm_pca,
        os.path.join(MODEL_DIR, f"{run_name}.joblib")
    )

    report = classification_report(
        y_test,
        svm_pred,
        labels=np.arange(len(class_names)),
        target_names=class_names,
        zero_division=0,
        output_dict=True
    )

    with open(
        os.path.join(RESULTS_DIR, f"{run_name}_classification_report.json"),
        "w",
        encoding="utf-8"
    ) as f:
        json.dump(report, f, indent=4)

    np.save(
        os.path.join(RESULTS_DIR, f"{run_name}_confusion_matrix.npy"),
        confusion_matrix(
            y_test, svm_pred, labels=np.arange(len(class_names))
        )
    )

    # --------------------------------------------------------
    # 6. Print the results
    # --------------------------------------------------------

    r = svm_results[run_name]
    print("Accuracy:", round(r["accuracy"] * 100, 2), "%")
    print("Macro precision:", round(r["macro_precision"], 4))
    print("Macro recall:", round(r["macro_recall"], 4))
    print("Macro F1-score:", round(r["macro_f1"], 4))
    print("Training time:", round(svm_train_time, 2), "seconds")

# Save results permanently.
result_path = os.path.join(RESULTS_DIR, "svm_pca_results.json")

with open(result_path, "w", encoding="utf-8") as f:
    json.dump(svm_results, f, indent=4)

print("\nAll SVM models trained successfully.")
print("Results saved to:", result_path)
print("Models saved in:", MODEL_DIR)
```

```python
# ============================================================
# SECTION 42: HYBRID SVM + KNN MODEL WITH PCA
# Method: Soft voting ensemble
# SVM and KNN each predict class probabilities. The hybrid
# averages them and picks the class with the highest average.
# Requires Sections 37 and 38 to be run first.
# ============================================================

import os
import json
import time
import joblib
import numpy as np

from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import VotingClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ------------------------------------------------------------
# 1. Settings
# ------------------------------------------------------------

# Feature sets to build a hybrid model for. Remove names from
# this list if you want a shorter run.
HYBRID_FEATURE_SETS = list(pca_data.keys())

# Weight of each model in the vote (1 : 1 = equal influence).
SVM_WEIGHT = 1
KNN_WEIGHT = 1

hybrid_results = {}

for feature_name in HYBRID_FEATURE_SETS:

    print("=" * 55)
    print("Training Hybrid SVM + KNN on", feature_name, "features WITH PCA")

    X_train_pca, X_test_pca = pca_data[feature_name]

    # Confirm that the PCA features and labels are aligned.
    assert X_train_pca.shape[0] == len(y_train)
    assert X_test_pca.shape[0] == len(y_test)

    # --------------------------------------------------------
    # 2. Define the two base models
    # probability=True lets the SVM output class probabilities
    # (this makes SVM training several times slower).
    # --------------------------------------------------------

    svm_model = SVC(
        kernel="rbf",
        C=10,
        gamma="scale",
        probability=True,
        random_state=42
    )

    knn_model = KNeighborsClassifier(
        n_neighbors=5,
        weights="distance",
        metric="euclidean",
        n_jobs=-1
    )

    # --------------------------------------------------------
    # 3. Combine them into one hybrid model
    # voting="soft" averages the predicted probabilities.
    # --------------------------------------------------------

    hybrid_model = VotingClassifier(
        estimators=[("svm", svm_model), ("knn", knn_model)],
        voting="soft",
        weights=[SVM_WEIGHT, KNN_WEIGHT]
    )

    # --------------------------------------------------------
    # 4. Train the hybrid model
    # --------------------------------------------------------

    start_time = time.time()
    hybrid_model.fit(X_train_pca, y_train)
    hybrid_train_time = time.time() - start_time

    # --------------------------------------------------------
    # 5. Predict the test set
    # --------------------------------------------------------

    start_time = time.time()
    hybrid_pred = hybrid_model.predict(X_test_pca)
    hybrid_predict_time = time.time() - start_time

    # --------------------------------------------------------
    # 6. Calculate the evaluation metrics
    # --------------------------------------------------------

    run_name = f"Hybrid_SVM_KNN_{feature_name}_PCA"

    hybrid_results[run_name] = {
        "feature_extraction": feature_name,
        "classifier": "Hybrid SVM + KNN",
        "dimensions_after_pca": int(X_train_pca.shape[1]),
        "variance_retained_percent": 95.0,
        "svm_weight": SVM_WEIGHT,
        "knn_weight": KNN_WEIGHT,
        "accuracy": float(accuracy_score(y_test, hybrid_pred)),
        "macro_precision": float(
            precision_score(y_test, hybrid_pred, average="macro",
                            zero_division=0)
        ),
        "macro_recall": float(
            recall_score(y_test, hybrid_pred, average="macro",
                         zero_division=0)
        ),
        "macro_f1": float(
            f1_score(y_test, hybrid_pred, average="macro",
                     zero_division=0)
        ),
        "training_time_seconds": float(hybrid_train_time),
        "prediction_time_seconds": float(hybrid_predict_time)
    }

    # --------------------------------------------------------
    # 7. Save the model, report and confusion matrix
    # --------------------------------------------------------

    joblib.dump(
        hybrid_model,
        os.path.join(MODEL_DIR, f"{run_name}.joblib")
    )

    report = classification_report(
        y_test,
        hybrid_pred,
        labels=np.arange(len(class_names)),
        target_names=class_names,
        zero_division=0,
        output_dict=True
    )

    with open(
        os.path.join(RESULTS_DIR, f"{run_name}_classification_report.json"),
        "w",
        encoding="utf-8"
    ) as f:
        json.dump(report, f, indent=4)

    np.save(
        os.path.join(RESULTS_DIR, f"{run_name}_confusion_matrix.npy"),
        confusion_matrix(
            y_test, hybrid_pred, labels=np.arange(len(class_names))
        )
    )

    # Save after every feature set so progress is not lost.
    with open(
        os.path.join(RESULTS_DIR, "hybrid_svm_knn_results.json"),
        "w",
        encoding="utf-8"
    ) as f:
        json.dump(hybrid_results, f, indent=4)

    # --------------------------------------------------------
    # 8. Print the results
    # --------------------------------------------------------

    r = hybrid_results[run_name]
    print("Accuracy:", round(r["accuracy"] * 100, 2), "%")
    print("Macro precision:", round(r["macro_precision"], 4))
    print("Macro recall:", round(r["macro_recall"], 4))
    print("Macro F1-score:", round(r["macro_f1"], 4))
    print("Training time:", round(hybrid_train_time, 2), "seconds")

print("\nAll hybrid SVM + KNN models trained successfully.")
print("Models saved in:", MODEL_DIR)
```

```python
# ============================================================
# SECTION 47: HYBRID SVM + KNN WITHOUT PCA
# Method: Soft voting ensemble (average of class probabilities)
# Input: Scaled original features
# Resume: Finished runs are skipped if you rerun this cell.
# ============================================================

import os
import json
import time
import joblib
import numpy as np

from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import VotingClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Weight of each model in the vote (1 : 1 = equal influence).
SVM_WEIGHT = 1
KNN_WEIGHT = 1

# Load earlier results so completed runs are not repeated.
result_path = os.path.join(RESULTS_DIR, "hybrid_svm_knn_no_pca_results.json")

if os.path.exists(result_path):
    with open(result_path, "r", encoding="utf-8") as f:
        hybrid_raw_results = json.load(f)
else:
    hybrid_raw_results = {}

for feature_name, (X_train_raw, X_test_raw) in raw_data.items():

    run_name = f"Hybrid_SVM_KNN_{feature_name}_NoPCA"

    if run_name in hybrid_raw_results:
        print("Skipping", run_name, "(already completed)")
        continue

    print("=" * 55)
    print("Training Hybrid SVM + KNN on", feature_name, "features WITHOUT PCA")

    # Confirm that the features and labels are aligned.
    assert X_train_raw.shape[0] == len(y_train)
    assert X_test_raw.shape[0] == len(y_test)

    # --------------------------------------------------------
    # 1. Define the two base models
    # probability=True lets the SVM output class probabilities
    # (this makes SVM training several times slower).
    # --------------------------------------------------------

    svm_model = SVC(
        kernel="rbf",
        C=10,
        gamma="scale",
        probability=True,
        random_state=42
    )

    knn_model = KNeighborsClassifier(
        n_neighbors=5,
        weights="distance",
        metric="euclidean",
        n_jobs=-1
    )

    # --------------------------------------------------------
    # 2. Combine them into one hybrid model
    # --------------------------------------------------------

    hybrid_raw = VotingClassifier(
        estimators=[("svm", svm_model), ("knn", knn_model)],
        voting="soft",
        weights=[SVM_WEIGHT, KNN_WEIGHT]
    )

    # --------------------------------------------------------
    # 3. Train the hybrid model
    # --------------------------------------------------------

    start_time = time.time()
    hybrid_raw.fit(X_train_raw, y_train)
    hybrid_train_time = time.time() - start_time

    # --------------------------------------------------------
    # 4. Predict the test set
    # --------------------------------------------------------

    start_time = time.time()
    hybrid_pred = hybrid_raw.predict(X_test_raw)
    hybrid_predict_time = time.time() - start_time

    # --------------------------------------------------------
    # 5. Calculate the evaluation metrics
    # --------------------------------------------------------

    hybrid_raw_results[run_name] = {
        "feature_extraction": feature_name,
        "classifier": "Hybrid SVM + KNN (no PCA)",
        "dimensions_used": int(X_train_raw.shape[1]),
        "svm_weight": SVM_WEIGHT,
        "knn_weight": KNN_WEIGHT,
        "accuracy": float(accuracy_score(y_test, hybrid_pred)),
        "macro_precision": float(
            precision_score(y_test, hybrid_pred, average="macro",
                            zero_division=0)
        ),
        "macro_recall": float(
            recall_score(y_test, hybrid_pred, average="macro",
                         zero_division=0)
        ),
        "macro_f1": float(
            f1_score(y_test, hybrid_pred, average="macro",
                     zero_division=0)
        ),
        "training_time_seconds": float(hybrid_train_time),
        "prediction_time_seconds": float(hybrid_predict_time)
    }

    # --------------------------------------------------------
    # 6. Save the model, report and confusion matrix
    # --------------------------------------------------------

    joblib.dump(
        hybrid_raw,
        os.path.join(MODEL_DIR, f"{run_name}.joblib")
    )

    report = classification_report(
        y_test,
        hybrid_pred,
        labels=np.arange(len(class_names)),
        target_names=class_names,
        zero_division=0,
        output_dict=True
    )

    with open(
        os.path.join(RESULTS_DIR, f"{run_name}_classification_report.json"),
        "w",
        encoding="utf-8"
    ) as f:
        json.dump(report, f, indent=4)

    np.save(
        os.path.join(RESULTS_DIR, f"{run_name}_confusion_matrix.npy"),
        confusion_matrix(
            y_test, hybrid_pred, labels=np.arange(len(class_names))
        )
    )

    # Save after every feature set so progress is not lost.
    with open(result_path, "w", encoding="utf-8") as f:
        json.dump(hybrid_raw_results, f, indent=4)

    # --------------------------------------------------------
    # 7. Print the results
    # --------------------------------------------------------

    r = hybrid_raw_results[run_name]
    print("Accuracy:", round(r["accuracy"] * 100, 2), "%")
    print("Macro precision:", round(r["macro_precision"], 4))
    print("Macro recall:", round(r["macro_recall"], 4))
    print("Macro F1-score:", round(r["macro_f1"], 4))
    print("Training time:", round(hybrid_train_time, 2), "seconds")

print("\nHybrid SVM + KNN runs without PCA finished.")
print("Results saved to:", result_path)
```

```python
# ============================================================
# SECTION 55: HYPERPARAMETER TUNING FOR SVM WITH PCA
# Feature set: ResNet50 + EfficientNetB0 hybrid features
# Pipeline: StandardScaler -> PCA -> SVM (RBF kernel)
# Tuned: PCA variance retained, C, gamma and class_weight
# Method: Randomized search with 3-fold cross-validation
# Requires Section 37 to be run first.
# ============================================================

import os
import json
import time
import joblib
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.model_selection import (
    RandomizedSearchCV,
    StratifiedKFold,
    train_test_split
)
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ------------------------------------------------------------
# 1. Build the original (not yet reduced) hybrid features
# ResNet50 (2048) + EfficientNetB0 (1280) = 3328 features.
# PCA is part of the pipeline, so it is fit separately inside
# every cross-validation fold (no information leaks).
# ------------------------------------------------------------

TUNE_FEATURE_SET = "Hybrid_Res_Eff"
parts = FEATURE_SETS[TUNE_FEATURE_SET]

X_train_raw = np.concatenate(
    [np.asarray(p[train_idx], dtype=np.float32) for p in parts], axis=1
)
X_test_raw = np.concatenate(
    [np.asarray(p[test_idx], dtype=np.float32) for p in parts], axis=1
)

assert X_train_raw.shape[0] == len(y_train)
assert X_test_raw.shape[0] == len(y_test)

n_features = X_train_raw.shape[1]
print("Feature set:", TUNE_FEATURE_SET)
print("Training shape:", X_train_raw.shape)
print("Testing shape:", X_test_raw.shape)

# ------------------------------------------------------------
# 2. Take a smaller stratified sample for the search
# Searching on all training data would take too long.
# The best settings are then used to train on ALL training data.
# The test set is NOT used for tuning.
# ------------------------------------------------------------

SUBSET_SIZE = 6000

X_search, _, y_search, _ = train_test_split(
    X_train_raw,
    y_train,
    train_size=min(SUBSET_SIZE, len(y_train) - 1),
    stratify=y_train,
    random_state=42
)

print("Samples used for the search:", len(y_search))

# ------------------------------------------------------------
# 3. Define the pipeline and the values to search
# After scaling, the total variance equals the number of
# features, so gamma="scale" is close to 1 / n_features.
# The gamma candidates are built around that value.
# ------------------------------------------------------------

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("pca", PCA(svd_solver="full", random_state=42)),
    ("svm", SVC(kernel="rbf", random_state=42)),
])

gamma_base = 1.0 / n_features

param_distributions = {
    # Fraction of variance kept by PCA (fewer or more components)
    "pca__n_components": [0.90, 0.95, 0.99],
    "svm__C": [1, 3, 10, 30, 100],
    "svm__gamma": [gamma_base * m for m in (0.3, 1.0, 3.0)],
    "svm__class_weight": [None, "balanced"],
}

# ------------------------------------------------------------
# 4. Run the randomized search with 3-fold cross-validation
# n_iter = number of random combinations to try.
# ------------------------------------------------------------

search = RandomizedSearchCV(
    estimator=pipeline,
    param_distributions=param_distributions,
    n_iter=15,
    scoring="accuracy",
    cv=StratifiedKFold(n_splits=3, shuffle=True, random_state=42),
    n_jobs=-1,
    verbose=2,
    random_state=42,
    refit=False
)

start_time = time.time()
search.fit(X_search, y_search)
search_time = time.time() - start_time

# ------------------------------------------------------------
# 5. Show and save the search results
# ------------------------------------------------------------

search_df = pd.DataFrame(search.cv_results_)
search_df = search_df[[
    "param_pca__n_components", "param_svm__C", "param_svm__gamma",
    "param_svm__class_weight", "mean_test_score", "std_test_score",
    "rank_test_score"
]].sort_values("rank_test_score")

display(search_df)
search_df.to_csv(
    os.path.join(RESULTS_DIR, f"svm_pca_tuning_search_{TUNE_FEATURE_SET}.csv"),
    index=False
)

# Average accuracy for each PCA setting (shows the effect of PCA).
print("\nAverage cross-validation accuracy by PCA variance retained:")
print(
    search_df.groupby("param_pca__n_components")["mean_test_score"]
    .mean().round(4)
)

best_params = search.best_params_
print("\nBest parameters:", best_params)
print("Best cross-validation accuracy:", round(search.best_score_ * 100, 2), "%")
print("Search time:", round(search_time / 60, 1), "minutes")

# ------------------------------------------------------------
# 6. Train the best pipeline on ALL training data
# ------------------------------------------------------------

best_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("pca", PCA(svd_solver="full", random_state=42)),
    ("svm", SVC(kernel="rbf", random_state=42)),
]).set_params(**best_params)

start_time = time.time()
best_pipeline.fit(X_train_raw, y_train)
train_time = time.time() - start_time

pca_dims = int(best_pipeline.named_steps["pca"].n_components_)
print("\nPCA components used:", pca_dims, "of", n_features)

# ------------------------------------------------------------
# 7. Evaluate once on the test set
# ------------------------------------------------------------

start_time = time.time()
y_pred = best_pipeline.predict(X_test_raw)
predict_time = time.time() - start_time

run_name = f"SVM_Tuned_{TUNE_FEATURE_SET}_PCA"

tuned_result = {
    "feature_extraction": TUNE_FEATURE_SET,
    "classifier": "SVM (tuned) with PCA",
    "best_params": {
        k: (float(v) if isinstance(v, (float, np.floating)) else
            int(v) if isinstance(v, (int, np.integer)) else v)
        for k, v in best_params.items()
    },
    "dimensions_after_pca": pca_dims,
    "cv_accuracy": float(search.best_score_),
    "accuracy": float(accuracy_score(y_test, y_pred)),
    "macro_precision": float(
        precision_score(y_test, y_pred, average="macro", zero_division=0)
    ),
    "macro_recall": float(
        recall_score(y_test, y_pred, average="macro", zero_division=0)
    ),
    "macro_f1": float(
        f1_score(y_test, y_pred, average="macro", zero_division=0)
    ),
    "training_time_seconds": float(train_time),
    "prediction_time_seconds": float(predict_time),
}

# ------------------------------------------------------------
# 8. Save the model, report, confusion matrix and results
# ------------------------------------------------------------

joblib.dump(best_pipeline, os.path.join(MODEL_DIR, f"{run_name}.joblib"))

report = classification_report(
    y_test, y_pred,
    labels=np.arange(len(class_names)),
    target_names=class_names,
    zero_division=0,
    output_dict=True
)
with open(
    os.path.join(RESULTS_DIR, f"{run_name}_classification_report.json"),
    "w", encoding="utf-8"
) as f:
    json.dump(report, f, indent=4)

np.save(
    os.path.join(RESULTS_DIR, f"{run_name}_confusion_matrix.npy"),
    confusion_matrix(y_test, y_pred, labels=np.arange(len(class_names)))
)

tuned_path = os.path.join(RESULTS_DIR, "tuned_results.json")
tuned_results = {}
if os.path.exists(tuned_path):
    with open(tuned_path, "r", encoding="utf-8") as f:
        tuned_results = json.load(f)

tuned_results[run_name] = tuned_result
with open(tuned_path, "w", encoding="utf-8") as f:
    json.dump(tuned_results, f, indent=4)

# ------------------------------------------------------------
# 9. Print the results
# ------------------------------------------------------------

print("\nTuned SVM test accuracy:", round(tuned_result["accuracy"] * 100, 2), "%")
print("Macro precision:", round(tuned_result["macro_precision"], 4))
print("Macro recall:", round(tuned_result["macro_recall"], 4))
print("Macro F1-score:", round(tuned_result["macro_f1"], 4))
print("Training time:", round(train_time, 2), "seconds")
```
