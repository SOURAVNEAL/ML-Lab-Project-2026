```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import mutual_info_classif, RFE, SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report
)
from xgboost import XGBClassifier
from catboost import CatBoostClassifier

RANDOM_STATE = 42
print("Libraries imported successfully.")
```

```python
uploaded = files.upload()

file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print("Loaded file:", file_name)
print("Dataset shape:", df.shape)
display(df.head())
```

```python
df.columns = df.columns.str.strip()

required = {"Date", "RainTomorrow"}
missing = required - set(df.columns)
if missing:
    raise ValueError(f"Required columns missing: {missing}")

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df["RainTomorrow"] = df["RainTomorrow"].astype("string").str.strip()
df = df.dropna(subset=["Date", "RainTomorrow"]).copy()
df = df[df["RainTomorrow"].isin(["Yes", "No"])].copy()

df["Target"] = df["RainTomorrow"].map({"No": 0, "Yes": 1}).astype(int)
df = df.sort_values("Date").reset_index(drop=True)

print("Cleaned dataset:", df.shape)
print(df["Target"].value_counts())
print((df["Target"].value_counts(normalize=True) * 100).round(2))
```

```python
# RISK_MM is intentionally excluded because it can reveal tomorrow's rainfall.
X_raw = df.drop(
    columns=[c for c in ["RainTomorrow", "Target", "RISK_MM"] if c in df.columns]
).copy()
y = df["Target"].copy()

X_raw["Year"] = df["Date"].dt.year
X_raw["Month"] = df["Date"].dt.month
X_raw["Day"] = df["Date"].dt.day
X_raw["DayOfYear"] = df["Date"].dt.dayofyear
X_raw["DayOfWeek"] = df["Date"].dt.dayofweek
X_raw = X_raw.drop(columns=["Date"])

print("Input features:", X_raw.shape)
print("Target:", y.shape)
display(X_raw.head())
```

```python
# Split by date so future records are not used to predict the past.
unique_dates = np.sort(df["Date"].unique())
n_dates = len(unique_dates)
train_cut = int(n_dates * 0.70)
valid_cut = int(n_dates * 0.85)

train_end = unique_dates[train_cut]
valid_end = unique_dates[valid_cut]

train_mask = df["Date"] < train_end
valid_mask = (df["Date"] >= train_end) & (df["Date"] < valid_end)
test_mask = df["Date"] >= valid_end

X_train_raw = X_raw.loc[train_mask].copy()
X_valid_raw = X_raw.loc[valid_mask].copy()
X_test_raw = X_raw.loc[test_mask].copy()

y_train = y.loc[train_mask].copy()
y_valid = y.loc[valid_mask].copy()
y_test = y.loc[test_mask].copy()

print("Train:", X_train_raw.shape, y_train.shape)
print("Validation:", X_valid_raw.shape, y_valid.shape)
print("Test:", X_test_raw.shape, y_test.shape)
print("Train last date:", df.loc[train_mask, "Date"].max())
print("Validation last date:", df.loc[valid_mask, "Date"].max())
print("Test first date:", df.loc[test_mask, "Date"].min())

for split_name, target in [("Train", y_train), ("Validation", y_valid), ("Test", y_test)]:
    print(split_name, "classes:", target.value_counts().to_dict())
    if target.nunique() != 2:
        raise ValueError(f"{split_name} split has only one class. Check the date split.")
```

```python
numeric_cols = X_train_raw.select_dtypes(include=np.number).columns.tolist()
categorical_cols = X_train_raw.select_dtypes(exclude=np.number).columns.tolist()

num_imputer = SimpleImputer(strategy="median")
cat_imputer = SimpleImputer(strategy="most_frequent")

def prepare_split(frame, fit=False):
    if fit:
        num_values = num_imputer.fit_transform(frame[numeric_cols])
        cat_values = cat_imputer.fit_transform(frame[categorical_cols]) if categorical_cols else np.empty((len(frame), 0))
    else:
        num_values = num_imputer.transform(frame[numeric_cols])
        cat_values = cat_imputer.transform(frame[categorical_cols]) if categorical_cols else np.empty((len(frame), 0))

    numeric_frame = pd.DataFrame(num_values, columns=numeric_cols, index=frame.index)
    categorical_frame = pd.DataFrame(cat_values, columns=categorical_cols, index=frame.index)
    combined = pd.concat([numeric_frame, categorical_frame], axis=1)
    return pd.get_dummies(combined, columns=categorical_cols, dtype=np.int8)

X_train = prepare_split(X_train_raw, fit=True)
X_valid = prepare_split(X_valid_raw)
X_test = prepare_split(X_test_raw)

feature_columns = X_train.columns.tolist()
X_valid = X_valid.reindex(columns=feature_columns, fill_value=0)
X_test = X_test.reindex(columns=feature_columns, fill_value=0)

X_train = X_train.replace([np.inf, -np.inf], np.nan).fillna(0)
X_valid = X_valid.replace([np.inf, -np.inf], np.nan).fillna(0)
X_test = X_test.replace([np.inf, -np.inf], np.nan).fillna(0)

print("Encoded shapes:", X_train.shape, X_valid.shape, X_test.shape)
```

```python
scaler = StandardScaler()
X_train_scaled = pd.DataFrame(
    scaler.fit_transform(X_train).astype(np.float32),
    columns=X_train.columns,
    index=X_train.index
)
print("Scaling completed.")
```

```python
sample_size = min(20000, len(X_train_scaled))
rng = np.random.RandomState(RANDOM_STATE)
sample_positions = np.sort(
    rng.choice(len(X_train_scaled), size=sample_size, replace=False)
)
X_fs = X_train_scaled.iloc[sample_positions].copy()
y_fs = y_train.iloc[sample_positions].copy()

print("Feature-selection sample:", X_fs.shape)
print(y_fs.value_counts())
```

```python
k = min(20, X_fs.shape[1])
selected_features = {}

# 1. Mutual Information
mi = mutual_info_classif(X_fs, y_fs, random_state=RANDOM_STATE)
selected_features["MI"] = (
    pd.Series(mi, index=X_fs.columns)
    .sort_values(ascending=False).head(k).index.tolist()
)

# 2. Recursive Feature Elimination
rfe_estimator = LogisticRegression(
    max_iter=1000, solver="liblinear",
    class_weight="balanced", random_state=RANDOM_STATE
)
rfe = RFE(estimator=rfe_estimator, n_features_to_select=k, step=0.2)
rfe.fit(X_fs, y_fs)
selected_features["RFE"] = X_fs.columns[rfe.support_].tolist()

# 3. SelectKBest
skb = SelectKBest(score_func=f_classif, k=k)
skb.fit(X_fs, y_fs)
selected_features["SelectKBest"] = X_fs.columns[skb.get_support()].tolist()

# 4. Greedy CFS-style correlation selection
corr_frame = X_fs.copy()
corr_frame["target"] = y_fs.to_numpy()
corr = corr_frame.corr().abs().fillna(0)
relevance = corr["target"].drop("target")
remaining = relevance.sort_values(ascending=False).index.tolist()
chosen = []

while remaining and len(chosen) < k:
    best_feature, best_score = None, -1
    for feature in remaining:
        candidate = chosen + [feature]
        rel = relevance[candidate].mean()
        if len(candidate) == 1:
            redundancy = 0
        else:
            sub = corr.loc[candidate, candidate].to_numpy()
            redundancy = (sub.sum() - len(candidate)) / (
                len(candidate) * (len(candidate) - 1)
            )
        score = rel / np.sqrt(1 + (len(candidate) - 1) * redundancy)
        if score > best_score:
            best_feature, best_score = feature, score
    chosen.append(best_feature)
    remaining.remove(best_feature)

selected_features["CFS"] = chosen

for method, feature_list in selected_features.items():
    print("\\n", method, ":", len(feature_list), "features")
    print(feature_list)
```

```python
for method, feature_list in selected_features.items():
    assert feature_list
    assert all(f in X_train.columns for f in feature_list)

print("All selected feature lists are valid.")
```

```python
boosting_configurations = [
    {"learning_rate": 0.01, "n_estimators": 500},
    {"learning_rate": 0.05, "n_estimators": 300},
    {"learning_rate": 0.10, "n_estimators": 200}
]
algorithms = ["XGBoost", "CatBoost", "Gradient Boosting"]
print(boosting_configurations)
```

```python
def create_model(algorithm, learning_rate, n_estimators):
    if algorithm == "XGBoost":
        return XGBClassifier(
            learning_rate=learning_rate,
            n_estimators=n_estimators,
            max_depth=5,
            subsample=0.8,
            colsample_bytree=0.8,
            eval_metric="logloss",
            tree_method="hist",
            random_state=RANDOM_STATE,
            n_jobs=-1
        )

    if algorithm == "CatBoost":
        return CatBoostClassifier(
            learning_rate=learning_rate,
            iterations=n_estimators,
            depth=6,
            loss_function="Logloss",
            verbose=False,
            random_seed=RANDOM_STATE,
            allow_writing_files=False
        )

    if algorithm == "Gradient Boosting":
        return GradientBoostingClassifier(
            learning_rate=learning_rate,
            n_estimators=n_estimators,
            max_depth=3,
            subsample=0.8,
            random_state=RANDOM_STATE
        )

    raise ValueError("Unknown algorithm")
print("Model function is ready.")
```

```python
results = []
validation_models = {}
experiment_number = 1

for fs_name, feature_list in selected_features.items():
    for algorithm in algorithms:
        for config in boosting_configurations:
            lr = config["learning_rate"]
            n_est = config["n_estimators"]

            print(
                "Experiment", experiment_number,
                "|", fs_name, "|", algorithm,
                "| learning_rate =", lr,
                "| n_estimators =", n_est
            )

            model = create_model(algorithm, lr, n_est)
            model.fit(X_train[feature_list], y_train)
            prediction = np.asarray(
                model.predict(X_valid[feature_list])
            ).reshape(-1)

            results.append({
                "Experiment": experiment_number,
                "Feature Selection": fs_name,
                "Algorithm": algorithm,
                "Learning Rate": lr,
                "n_estimators": n_est,
                "Accuracy": accuracy_score(y_valid, prediction) * 100,
                "Precision": precision_score(y_valid, prediction, zero_division=0) * 100,
                "Recall": recall_score(y_valid, prediction, zero_division=0) * 100,
                "F1-Score": f1_score(y_valid, prediction, zero_division=0) * 100
            })

            validation_models[experiment_number] = {
                "features": feature_list,
                "algorithm": algorithm,
                "feature_selection": fs_name
            }
            experiment_number += 1

results_df = pd.DataFrame(results)
print("Total experiments:", len(results_df))
display(results_df.round(3))
```

```python
final_results = results_df.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

final_results.insert(
    0,
    "Rank",
    range(1, len(final_results) + 1)
)

print("========== FINAL RANKING BASED ON ACCURACY ==========")

display(
    final_results.round(4)
)

print("\nBEST EXPERIMENT BASED ON ACCURACY")

display(
    final_results.head(1).round(4)
)
```

```python
best_experiment = final_results.iloc[0]

print("========== BEST MODEL ==========")

print("Experiment:", best_experiment["Experiment"])
print("Accuracy:", best_experiment["Accuracy"])
print("Precision:", best_experiment["Precision"])
print("Recall:", best_experiment["Recall"])
print("F1-Score:", best_experiment["F1-Score"])
```

```python
best_row = final_results.iloc[0]

best_experiment = int(best_row["Experiment"])
best_info = validation_models[best_experiment]
best_features = best_info["features"]

X_train_valid = pd.concat(
    [X_train, X_valid], axis=0
)

y_train_valid = pd.concat(
    [y_train, y_valid], axis=0
)

final_model = create_model(
    best_info["algorithm"],
    float(best_row["Learning Rate"]),
    int(best_row["n_estimators"])
)

final_model.fit(
    X_train_valid[best_features],
    y_train_valid
)

print("========== BEST MODEL DETAILS ==========")
print("Best Experiment:", best_experiment)
print("Feature Selection:", best_info["feature_selection"])
print("Algorithm:", best_info["algorithm"])
print("Learning Rate:", best_row["Learning Rate"])
print("Estimators:", best_row["n_estimators"])
print("Validation Accuracy:", round(best_row["Accuracy"], 2), "%")
print("Selected Features:", len(best_features))
```

```python
test_predictions = np.asarray(
    final_model.predict(X_test[best_features])
).reshape(-1)

test_accuracy = accuracy_score(
    y_test, test_predictions
) * 100

test_precision = precision_score(
    y_test, test_predictions,
    zero_division=0
) * 100

test_recall = recall_score(
    y_test, test_predictions,
    zero_division=0
) * 100

test_f1 = f1_score(
    y_test, test_predictions,
    zero_division=0
) * 100

print("========== FINAL TEST PERFORMANCE ==========")

print(f"Accuracy  : {test_accuracy:.2f}%")
print(f"Precision : {test_precision:.2f}%")
print(f"Recall    : {test_recall:.2f}%")
print(f"F1-Score  : {test_f1:.2f}%")

print("\nCLASSIFICATION REPORT")

print(
    classification_report(
        y_test,
        test_predictions,
        zero_division=0
    )
)
```
